1 — Import libraries and load dataset

```python
# 1. Import Libraries
import pandas as pd
import os

# 2. Load Original Dataset
input_file = r"C:\Users\dell\Downloads\online+retail\Online Retail.xlsx"

df = pd.read_excel(input_file)

# Display first 5 rows
print("First 5 rows:")
display(df.head())

# Dataset shape
print("Original Dataset Shape:", df.shape)

# Column names and data types
print("\nDataset Information:")
df.info()
```

2 — Data quality inspection

```python
# 1. Check Missing Values
print("Missing Values:")
print(df.isnull().sum())

# 2. Check Duplicate Records
print("\nDuplicate Rows:", df.duplicated().sum())

# 3. Check Data Types
print("\nData Types:")
print(df.dtypes)

# 4. Check Unique Countries
print("\nUnique Countries:", df["Country"].nunique())

# 5. Check Descriptions with Extra Spaces
extra_spaces = (
    df["Description"].notna()
    & df["Description"].ne(df["Description"].str.strip())
).sum()

print("\nDescriptions with Extra Spaces:", extra_spaces)

# 6. Check Negative Quantity
print("\nNegative Quantity Records:", (df["Quantity"] < 0).sum())

# 7. Check Negative Unit Price
print("Negative UnitPrice Records:", (df["UnitPrice"] < 0).sum())

# 8. Check Cancelled Invoices
cancelled_invoices = df["InvoiceNo"].astype(str).str.startswith("C").sum()

print("Cancelled Invoice Records:", cancelled_invoices)
```

3 — Data cleaning

```python
# Create a copy of the original dataset
df_clean = df.copy()

# 1. Remove Duplicate Records
df_clean = df_clean.drop_duplicates()

# 2. Handle Missing Description Values
df_clean["Description"] = df_clean["Description"].fillna("Unknown")

# 3. Remove Extra Spaces from Description
df_clean["Description"] = df_clean["Description"].str.strip()

# 4. Convert CustomerID to Nullable Integer
df_clean["CustomerID"] = df_clean["CustomerID"].astype("Int64")

print("Data Cleaning Completed!")
```

4 — Final validation and export

```python
# 1. Final Dataset Shape
print("Final Dataset Shape:", df_clean.shape)

# 2. Check Missing Values
print("\nMissing Values After Cleaning:")
print(df_clean.isnull().sum())

# 3. Check Duplicate Records
print("\nDuplicate Rows After Cleaning:", df_clean.duplicated().sum())

# 4. Check Final Data Types
print("\nFinal Data Types:")
print(df_clean.dtypes)

# 5. Create Project Folder
project_folder = r"C:\Users\dell\Downloads\Data-Cleaning-Internship-Task-1"

os.makedirs(project_folder, exist_ok=True)

# 6. Export Cleaned Dataset
output_file = os.path.join(
    project_folder,
    "cleaned_online_retail.csv"
)

df_clean.to_csv(output_file, index=False)

print("\nCleaned dataset exported successfully!")
print("Saved Location:", output_file)
```

```python
import gzip
import os

input_file = os.path.join(
    project_folder,
    "cleaned_online_retail.csv"
)

output_file = os.path.join(
    project_folder,
    "cleaned_online_retail.csv.gz"
)

with open(input_file, "rb") as f_in:
    with gzip.open(output_file, "wb") as f_out:
        f_out.writelines(f_in)

print("Compressed file created successfully!")
print("Compressed size:", round(os.path.getsize(output_file) / (1024 * 1024), 2), "MB")
```
